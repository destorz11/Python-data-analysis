# Лабораторная работа №1. Иерархический анализ

## 1. Описание данных

**Тема:** экономика городов.

Данные описывают экономические условия в 48 городах мира в 1991 году. Их собрал отдел экономических исследований банка Union (Швейцария).

- **Файл:** `data/Econom_Cities_data.csv` (разделитель `;`, десятичная запятая)
- **Число наблюдений:** 48

| Переменная | Название | Описание |
|---|---|---|
| `City` | Город | Название города |
| `Work` | Работа | Взвешенное среднее числа рабочих часов, сосчитанное по 12 профессиям |
| `Price` | Цена | Индекс цен 112 товаров и услуг, включая арендную плату за жильё (значение для Цюриха взято за 100%) |
| `Salary` | Заработная плата | Индекс заработной платы за час работы, сосчитанный по 12 профессиям после налогов и вычетов (значение для Цюриха взято за 100%) |


## 2. Постановка задачи

С помощью иерархического кластерного анализа разбить города на группы с похожими экономическими условиями по признакам `Work`, `Price` и `Salary`, выбрать число кластеров и описать получившиеся группы.

## Этап 2. Обработка пропусков

### 2.1. Замена -9999 на NaN

Заменю `-9999` на `NaN`, чтобы эти значения не считались обычными числами. Таблицу `df` не меняю, работаю с копией.

```python
numeric_columns = ['Work', 'Price', 'Salary']

df_with_nan = df.copy()
df_with_nan[numeric_columns] = df_with_nan[numeric_columns].replace(-9999, np.nan)

df_with_nan.isna().sum().to_frame('Пропуски')
```

### 2.2. Удаление городов с пропусками

У Cairo и Jakarta нет двух признаков из трёх. Если заполнить их средним или медианой, расстояния между городами исказятся, поэтому эти города в анализ брать не буду.

```python
excluded_cities = df_with_nan[df_with_nan[numeric_columns].isna().any(axis=1)]

df_clean = df_with_nan.dropna(subset=numeric_columns).reset_index(drop=True)
df_clean['Work'] = df_clean['Work'].astype(int)

excluded_cities
```

### 2.3. Проверка результата

```python
print(f'Было строк: {df.shape[0]}, удалено: {excluded_cities.shape[0]}, осталось: {df_clean.shape[0]}')
print('Пропусков (NaN):', df_clean.isna().sum().sum())
print('Значений -9999:', df_clean[numeric_columns].eq(-9999).sum().sum())

display(df_clean.dtypes.to_frame('Тип'))
df_clean[numeric_columns].agg(['min', 'max'])
```

### Итог этапа 2

После замены `-9999` на `NaN` и удаления Cairo и Jakarta в `df_clean` осталось 46 городов без пропусков. Дальше работаю с `df_clean`.

## Этап 3. Разведочный анализ

### 3.1. Описательные статистики

### 3.2. Распределения признаков

### 3.3. Матрица корреляций

Посчитаю корреляцию Пирсона между признаками.

```python
correlation = df_clean[numeric_columns].corr()

plt.imshow(correlation, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar()
plt.xticks(range(3), numeric_columns)
plt.yticks(range(3), numeric_columns)

for i in range(3):
    for j in range(3):
        plt.text(j, i, round(correlation.iloc[i, j], 2), ha='center', va='center')

plt.title('Корреляция признаков')
plt.show()
```

### 3.4. Поиск выбросов

Выбросы ищу по правилу 1.5 IQR: выброс это значение меньше $Q_1 - 1.5 \cdot IQR$ или больше $Q_3 + 1.5 \cdot IQR$, где $IQR = Q_3 - Q_1$.

```python
q1 = df_clean[numeric_columns].quantile(0.25)
q3 = df_clean[numeric_columns].quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

for column in numeric_columns:
    mask = (df_clean[column] < lower[column]) | (df_clean[column] > upper[column])
    print(column, 'границы:', round(lower[column], 2), round(upper[column], 2))
    print('выбросы:', list(df_clean.loc[mask, 'City']))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(13, 4))

for ax, column in zip(axes, numeric_columns):
    ax.boxplot(df_clean[column])
    ax.set_title(column)
    ax.set_xticks([])

plt.tight_layout()
plt.show()
```

### Итог этапа 3

Разброс у признаков разный: стандартное отклонение `Work` 174.34, а у `Price` и `Salary` 21.39 и 24.76. Поэтому перед расчетом расстояний признаки надо стандартизировать.

`Work` скошен вправо, у `Price` и `Salary` асимметрия меньше. На гистограмме `Salary` много городов с низким индексом: 10 из 46 попали в первый интервал, возможно, это отдельная группа.

`Price` и `Salary`,что  логично - связаны. Чем больше `Work`, тем в среднем ниже цены и зарплаты.

Выброс: Hong_Kong по `Work`.

## Этап 4. Стандартизация признаков

Значения `Work` порядка двух тысяч часов, а `Price` и `Salary` это индексы примерно до 100. Если считать расстояния по исходным данным, почти все будет определять `Work`. Поэтому привожу признаки к z-оценкам: $z = (x - \bar{x}) / s$.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X = scaler.fit_transform(df_clean[numeric_columns])

df_scaled = pd.DataFrame(X, columns=numeric_columns, index=df_clean['City'])
df_scaled.head()
```

Проверю, что средние стали равны 0, а стандартные отклонения 1.

```python
print('Средние:')
print(df_scaled.mean().round(3))
print()
print('Стандартные отклонения:')
print(df_scaled.std(ddof=0).round(3))
```

### Итог этапа 4

Признаки теперь в одном масштабе: средние равны 0, стандартные отклонения 1. Дальше для расчета расстояний использую `df_scaled`.

## Этап 5. Матрица расстояний

Считаю евклидовы расстояния между городами по стандартизованным признакам.

```python
from scipy.spatial.distance import pdist, squareform

distances = pdist(df_scaled, metric='euclidean')

dist_matrix = pd.DataFrame(
    squareform(distances), index=df_scaled.index, columns=df_scaled.index
)

print('Размер матрицы:', dist_matrix.shape)
dist_matrix.iloc[:5, :5].round(2)
```

Посмотрю, какие города ближе всего друг к другу, а какие дальше всего.

```python
upper_triangle = np.triu(np.ones(dist_matrix.shape, dtype=bool), k=1)
pairs = dist_matrix.where(upper_triangle).stack().sort_values()

print('Самые близкие пары:')
print(pairs.head(5).round(3))
print()
print('Самые далекие пары:')
print(pairs.tail(5).round(3))
```

### Итог этапа 5

Ближе всего друг к другу London и Paris. Самая далекая пара это Manila и Oslo, причем Oslo есть в четырех из пяти самых далеких пар. Дальше по этой матрице строю иерархию.

## Этап 6. Выбор метода связи

По совету Вадима Леонардовича для новичков использую метод Уорда (`ward`). Он объединяет кластеры так, чтобы разброс внутри них рос как можно меньше, поэтому кластеры получаются компактными.

## Этап 7. Построение иерархии

Строю иерархию методом `ward`. Использую евклидовы расстояния из этапа 5.

```python
from scipy.cluster.hierarchy import dendrogram, fcluster, linkage
from sklearn.metrics import silhouette_score

Z = linkage(distances, method='ward')

print('Объединений:', Z.shape[0])
print('Высота последнего объединения:', round(Z[-1, 2], 2))
```

Построю дендрограмму.

```python
plt.figure(figsize=(14, 6))
dendrogram(Z, labels=list(df_scaled.index), leaf_rotation=90, leaf_font_size=9)
plt.title('Дендрограмма, метод ward')
plt.ylabel('Расстояние объединения')
plt.show()
```

### Итог этапа 7

Получилось 45 объединений, последнее на высоте 12.4. На дендрограмме видны две большие ветви, и внутри каждой из них есть более мелкие группы.

## Этап 8. Выбор числа кластеров

### 8.1. Каменистая осыпь

Выведу расстояния объединения для последних шагов и посмотрю, на сколько они меняются.

```python
heights = Z[:, 2][::-1][:10]

scree = pd.DataFrame({'Расстояние': heights}, index=range(1, 11))
scree.index.name = 'Число кластеров'
scree['Скачок'] = scree['Расстояние'] - scree['Расстояние'].shift(-1)

scree.round(3)
```

```python
plt.figure(figsize=(8, 4))
plt.plot(scree.index, scree['Расстояние'], marker='o')
plt.title('Каменистая осыпь: ward')
plt.xlabel('Число кластеров')
plt.ylabel('Расстояние объединения')
plt.show()
```

### 8.2. Критерий качества Silhouette

Silhouette меняется от -1 до 1. Чем больше значение, тем лучше кластеры отделены друг от друга. Посчитаю его для числа кластеров от 2 до 10.

```python
k_values = range(2, 11)
scores = []

for k in k_values:
    labels = fcluster(Z, t=k, criterion='maxclust')
    scores.append(silhouette_score(df_scaled, labels))

plt.figure(figsize=(8, 4))
plt.plot(k_values, scores, marker='o')
plt.title('Silhouette: ward')
plt.xlabel('Число кластеров')
plt.ylabel('Silhouette')
plt.show()

pd.Series(scores, index=k_values, name='Silhouette').round(3).to_frame()
```

### 8.3. Выбор числа кластеров

На осыпи самый большой скачок при переходе к 1 кластеру. Следующий заметный скачок при переходе от 4 кластеров к 3, а при переходе от 3 к 2 он маленький. Значит, останавливаться можно на 2 или на 4 кластерах.

Построю дендрограмму с разрезом на 4 кластера.

```python
n_clusters = 4
cut = (Z[-4, 2] + Z[-3, 2]) / 2

plt.figure(figsize=(14, 6))
dendrogram(Z, labels=list(df_scaled.index), leaf_rotation=90, leaf_font_size=9,
           color_threshold=cut, above_threshold_color='gray')
plt.axhline(cut, color='black', linestyle='--', label='разрез')
plt.title('Дендрограмма ward, разрез на 4 кластера')
plt.ylabel('Расстояние объединения')
plt.legend()
plt.show()
```

## Этап 9. Формирование кластеров

Разрезаю дерево на 4 кластера и записываю номер кластера для каждого города.

```python
df_clusters = df_clean.copy()
df_clusters['Cluster'] = fcluster(Z, t=n_clusters, criterion='maxclust')

df_clusters['Cluster'].value_counts().sort_index().to_frame('Число городов')
```

Посмотрю, какие города попали в каждый кластер.

```python
for cluster, group in df_clusters.groupby('Cluster'):
    print(f'Кластер {cluster}, городов: {len(group)}')
    print(', '.join(group['City']))
    print()
```

Проверю, что кластер есть у всех городов, и посчитаю Silhouette для итогового разбиения.

```python
print('Всего городов:', len(df_clusters))
print('Городов без кластера:', df_clusters['Cluster'].isna().sum())
print('Silhouette:', round(silhouette_score(df_scaled, df_clusters['Cluster']), 3))
```

### Итог этапа 9

Получилось 4 кластера: 6, 19, 7 и 14 городов

## Этап 10. Интерпретация

### 10.1. Средние значения по кластерам

Считаю средние по исходным данным, чтобы их было удобно сравнивать.

```python
profile = df_clusters.groupby('Cluster')[numeric_columns].mean().round(2)
profile['Городов'] = df_clusters['Cluster'].value_counts()

profile
```

Диапазоны значений внутри кластеров:

```python
df_clusters.groupby('Cluster')[numeric_columns].agg(['min', 'max'])
```

Дам кластерам названия. В первом самые высокие зарплаты и цены, поэтому это «Элитки». Во втором зарплаты и цены тоже высокие, но ниже, поэтому это «Богачи». В третьем низкие цены и зарплаты при обычном рабочем времени, поэтому это «Бюджетные города». В четвертом работают дольше всех за самую низкую зарплату, поэтому это «Работяги».

```python
cluster_names = {1: 'Элитки', 2: 'Богачи', 3: 'Бюджетные города', 4: 'Работяги'}

df_clusters['Название'] = df_clusters['Cluster'].map(cluster_names)
profile['Название'] = profile.index.map(cluster_names)

profile
```

### 10.2. Графики

```python
fig, axes = plt.subplots(1, 3, figsize=(13, 4))

for ax, column in zip(axes, numeric_columns):
    ax.bar(profile['Название'], profile[column])
    ax.set_title(f'Среднее {column}')
    ax.tick_params(axis='x', rotation=40)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for cluster, group in df_clusters.groupby('Cluster'):
    axes[0].scatter(group['Price'], group['Salary'], label=cluster_names[cluster])
    axes[1].scatter(group['Work'], group['Salary'], label=cluster_names[cluster])

axes[0].set_xlabel('Price')
axes[0].set_ylabel('Salary')
axes[0].legend()

axes[1].set_xlabel('Work')
axes[1].set_ylabel('Salary')
axes[1].legend()

plt.tight_layout()
plt.show()
```

Построю еще 3D график, чтобы посмотреть на все три признака сразу. Его можно вращать мышью, а при наведении на точку видно название города.

```python
import plotly.express as px

fig_3d = px.scatter_3d(
    df_clusters,
    x='Work',
    y='Price',
    z='Salary',
    color='Название',
    hover_name='City',
    category_orders={'Название': ['Элитки', 'Богачи', 'Бюджетные города', 'Работяги']},
    title='Города по трем признакам',
    height=650,
)
fig_3d.update_traces(marker_size=5)
fig_3d.show()
```

### 10.3. Описание кластеров

Кластер 1, «Элитки» (Geneva, Helsinki, Oslo, Stockholm, Tokyo, Zurich). Самые высокие цены и зарплаты: в среднем `Price` 108.55, `Salary` 71.30, `Work` 1780.5.

Кластер 2, «Богачи» (19 городов). Цены и зарплаты тоже высокие, но ниже, а рабочих часов почти столько же. Это города Европы и Северной Америки и еще Sydney.

Кластер 3, «Бюджетные города» (Athens, Lagos, Lisbon, Nicosia, Rio_de_Janeiro, San_Paulo, Seoul). Низкие цены и зарплаты, а рабочих часов примерно как в богатых городах.

Кластер 4, «Работяги» (14 городов). Больше всего рабочих часов, самая низкая зарплата и низкие цены. Это города Азии, Латинской Америки и Африки и еще Tel_Aviv.

По диапазонам видно, что «Элитки» и «Богачи» разделяются по `Price`, а «Бюджетные города» и «Работяги» по `Work`

### 10.4. Что будет при 2 кластерах

Посмотрю, что изменится, если взять 2 кластера. Silhouette там самый высокий среди небольших k (0.470).

```python
df_clusters['Cluster_2'] = fcluster(Z, t=2, criterion='maxclust')

profile_2 = df_clusters.groupby('Cluster_2')[numeric_columns].mean().round(2)
profile_2['Городов'] = df_clusters['Cluster_2'].value_counts()

profile_2
```

Сравню с разбиением на 4 кластера.

```python
pd.crosstab(df_clusters['Название'], df_clusters['Cluster_2'])
```

При k = 2 «Элитки» и «Богачи» объединяются в одну группу из 25 городов, а «Бюджетные города» и «Работяги» в другую из 21 города (52.40 и 16.35). Самый большой скачок на осыпи как раз при переходе от 2 кластеров к 1, то есть главное различие в данных идет между богатыми и небогатыми городами.

Но при таком делении теряется разница между 6 самыми дорогими городами и остальными богатыми, а также между «Работягами» и «Бюджетными городами», которые различаются рабочим временем.

### 10.5. Покупательная способность

`Price` и `Salary` связаны, поэтому в расстояниях они почти дублируют друг друга. Проверю, изменятся ли кластеры, если заменить их одним признаком: $RealWage = Salary / Price \cdot 100$. Чем он больше, тем больше можно купить на зарплату в этом городе.

```python
df_real = df_clean.set_index('City').copy()
df_real['RealWage'] = df_real['Salary'] / df_real['Price'] * 100

display(df_real['RealWage'].sort_values(ascending=False).head().round(1).to_frame('Самые высокие'))
df_real['RealWage'].sort_values().head().round(1).to_frame('Самые низкие')
```

Заново построю иерархию методом `ward`, но по признакам `Work` и `RealWage`.

```python
X_real = StandardScaler().fit_transform(df_real[['Work', 'RealWage']])
Z_real = linkage(X_real, method='ward')

scores_real = []
for k in range(2, 7):
    labels = fcluster(Z_real, t=k, criterion='maxclust')
    scores_real.append(silhouette_score(X_real, labels))

pd.Series(scores_real, index=range(2, 7), name='Silhouette').round(3).to_frame()
```

```python
df_clusters['Cluster_real'] = fcluster(Z_real, t=4, criterion='maxclust')

pd.crosstab(df_clusters['Название'], df_clusters['Cluster_real'])
```

Покажу оба разбиения на одном графике.

```python
df_clusters['RealWage'] = df_real['RealWage'].values

fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharex=True, sharey=True)

for cluster, group in df_clusters.groupby('Cluster'):
    axes[0].scatter(group['Work'], group['RealWage'], label=cluster_names[cluster])

for cluster, group in df_clusters.groupby('Cluster_real'):
    axes[1].scatter(group['Work'], group['RealWage'], label=f'Группа {cluster}')

axes[0].set_title('Work, Price, Salary')
axes[1].set_title('Work и RealWage')

for ax in axes:
    ax.set_xlabel('Work')
    ax.set_ylabel('RealWage')
    ax.legend()

plt.tight_layout()
plt.show()
```

Дендрограмма для варианта с `RealWage`, разрез на 4 кластера.

Получается, что группы с низкими зарплатами устойчивые, а деление богатых городов зависит от набора признаков.
